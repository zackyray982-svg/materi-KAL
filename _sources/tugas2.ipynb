## 1. MINOR

Menentukan minor setiap elemen matriks sebagai dasar untuk perhitungan kofaktor, determinan, dan invers

Minor
$M_{ij}$
adalah determinan submatriks yang diperoleh dengan menghapus baris ke-i dan kolom ke-j dari matriks A.

Diketahui

$A =
\begin{bmatrix}
a & b & c & d & e \\
f & g & h & i & j \\
k & l & m & n & o \\
p & q & r & s & t \\
u & v & w & x & y
\end{bmatrix}$

\end{bmatrix}$

$M_{12} =

## 2. MENENTUKAN KOFAKTOR MATRIKS

Menentukan nilai kofaktor dari setiap elemen matriks berdasarkan minor yang telah diperoleh pada langkah sebelumnya

Kofaktor $C_{ij}$ adalah hasil perkalian antara minor $M_{ij}$ dengan tanda $(-1)^{i+j}$:

$$
C_{ij} = (-1)^{i+j} \cdot M_{ij}$$

+ Jika $i$+$j$
genap → tanda positif (+)

+ Jika $i$+$j$
ganjil → tanda negatif (−)

## Tanda Kofaktor

$$\begin{bmatrix}
+ & - & + & - & + \\
- & + & - & + & - \\
+ & - & + & - & + \\
- & + & - & + & - \\
+ & - & + & - & +
\end{bmatrix}
$$


### Contoh baris pertama:

$C_{11}=+M_{11}$

$C_{12}=-M_{12}$

$C_{13}=+M_{13}$

$C_{14}=-M_{14}$

$C_{15}=+M_{15}$
	​


Setelah semua kofaktor $C_{ij}$ diperoleh, nilai-nilai tersebut disusun sesuai posisi elemen aslinya dalam matriks

$$
C =
\begin{bmatrix}
C_{11} & C_{12} & C_{13} & C_{14} & C_{15} \\
C_{21} & C_{22} & C_{23} & C_{24} & C_{25} \\
C_{31} & C_{32} & C_{33} & C_{34} & C_{35} \\
C_{41} & C_{42} & C_{43} & C_{44} & C_{45} \\
C_{51} & C_{52} & C_{53} & C_{54} & C_{55}
\end{bmatrix}
$$

## 3. DETERMINAN $det(A)$

Determinan matriks 5×5 dihitung dengan ekspansi baris pertama:

$\det(A)= aC_{11}+bC_{12}+cC_{13}+dC_{14}+eC_{15}$

Substitusi kofaktor:

$\det(A)=
aM_{11}
-bM_{12}
+cM_{13}
-dM_{14}
+eM_{15}$

Artinya determinan adalah jumlah dari elemen baris pertama dikalikan kofaktornya

## 4. ADJOIN $adj(A)$

$Adj(A) = C^T$

Transpose berarti:

+ baris menjadi kolom
+ kolom menjadi baris

Bentuk Adjoin Matriks

$$Adj(A) =
\begin{bmatrix}
C_{11} & C_{21} & C_{31} & C_{41} & C_{51} \\
C_{12} & C_{22} & C_{32} & C_{42} & C_{52} \\
C_{13} & C_{23} & C_{33} & C_{43} & C_{53} \\
C_{14} & C_{24} & C_{34} & C_{44} & C_{54} \\
C_{15} & C_{25} & C_{35} & C_{45} & C_{55}
\end{bmatrix}$$

##5. INVERS MATRIKS $A^{-1}$

Rumus invers matriks:

$A^{-1} = \frac{1}{\det(A)} \, \text{adj}(A)$

Substitusi:

$A^{-1} =
\frac{1}{\det(A)}
\begin{bmatrix}
C_{11} & C_{21} & C_{31} & C_{41} & C_{51} \\
C_{12} & C_{22} & C_{32} & C_{42} & C_{52} \\
C_{13} & C_{23} & C_{33} & C_{43} & C_{53} \\
C_{14} & C_{24} & C_{34} & C_{44} & C_{54} \\
C_{15} & C_{25} & C_{35} & C_{45} & C_{55}
\end{bmatrix}$

## 2. Menentukan Determinan Matriks

Menggunakan ekspansi kofaktor pada baris pertama:

$$
\det(A) = 1 \cdot \det(A_{11}) - 2 \cdot \det(A_{12})
$$

### a. Menghitung $A_{11}$

$$
A_{11} =
\begin{bmatrix}
1 & 3 & 0 & 0 \\
0 & 1 & 4 & 0 \\
0 & 0 & 1 & 5 \\
0 & 1 & 0 & 0
\end{bmatrix}
$$


$$
= 1
\begin{vmatrix}
1 & 4 & 0 \\
0 & 1 & 5 \\
1 & 0 & 0
\end{vmatrix}
$$

$$
= 1[1(1\cdot0 - 5\cdot0) - 4(0\cdot0 - 5\cdot1)]
$$

$$
= 1[0 - 4(-5)] = 20
$$

### Detail Perhitungan Determinan 3x3

$$
\begin{vmatrix}
1 & 4 & 0 \\
0 & 1 & 5 \\
1 & 0 & 0
\end{vmatrix}
$$

$$
= 1(1\cdot0 - 5\cdot0)
- 4(0\cdot0 - 5\cdot1)
+ 0(0\cdot0 - 1\cdot1)
$$

$$
= 0 - 4(-5) + 0
$$

$$
= 20
$$

### b. Menghitung $A_{12}$

$$
A_{12} =
\begin{bmatrix}
0 & 3 & 0 & 0 \\
0 & 1 & 4 & 0 \\
0 & 0 & 1 & 5 \\
1 & 1 & 0 & 0
\end{bmatrix}
$$

$$
=
\begin{vmatrix}
3 & 0 & 0 \\
1 & 4 & 0 \\
0 & 1 & 5
\end{vmatrix}
$$

$$
= 3(4\cdot5 - 0\cdot1) = 60
$$



### c. Hasil Determinan

$$
\det(A) = 1(20) - 2(60)
$$

$$
\det(A) = 20 - 60 = -40
$$
