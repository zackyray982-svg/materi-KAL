$$\sum_{k=1}^n (-1)^{i+k} a_{ik} M_{ik}$$


dengan $M_{ij}$ adalah minior dari matrik A dan

$$ M_{ij} = \det A_{ij}. $$


$A_{ij}$ adalah submatrik dengan menghapus baris i dan kolom kolom j dari matrix $A_{mxn}$ dengan $1 \le i, j \le n$





1. $$A = \begin{bmatrix} -7 & -5 \\ 1 & 4 \end{bmatrix}$$




2. $$ A = \begin{bmatrix} 0 & 2 & -3 \\ 1 & -2 & -1 \\ 0 & 0 & 1 \end{bmatrix}$$

### JAWABAN


## A. Determinan Matriks (Ekspansi Baris)

Determinan matriks dapat dihitung dengan ekspansi baris:

$$
\det(A) = \sum_{k=1}^{n} (-1)^{i+k} a_{ik} M_{ik}
$$

dengan:

* $M_{ik}$ = minor (determinan submatriks hasil menghapus baris $i$ dan kolom $k$)

* $C_{ik} = (-1)^{i+k} M_{ik}$ = kofaktor

---

## 1. Matriks 2×2

Diketahui:
$$
A =
\begin{bmatrix}
-7 & -5 \\
1 & 4
\end{bmatrix}
$$

Ekspansi baris pertama:

$$
\det(A) = (-1)^{1+1}(-7)M_{11} + (-1)^{1+2}(-5)M_{12}
$$

Hitung minor:

$$
M_{11} =
\begin{vmatrix}
4
\end{vmatrix} = 4
$$

$$
M_{12} =
\begin{vmatrix}
1
\end{vmatrix} = 1
$$

Hitung kofaktor:

$$
C_{11} = (+1)(4) = 4, \quad C_{12} = (-1)(1) = -1
$$

Substitusi:

$$
\det(A) = (-7)(4) + (-5)(-1)
$$

$$
= -28 + 5 = -23
$$

**Hasil:**
$$
\det(A) = -23
$$

## 2. Matriks 3×3

$$
\det(A) = \sum_{k=1}^{3} (-1)^{1+k} a_{1k} M_{1k}

$$

$$

\begin{vmatrix}

\end{vmatrix}

$$

$$
M_{12} =
\begin{vmatrix}

0 & 1
\end{vmatrix}
= (1)(1) - (-1)(0)
= 1
$$

$$
M_{13} =
\begin{vmatrix}
1 & -2 \\
0 & 0
\end{vmatrix}
= (1)(0) - (-2)(0)
= 0
$$


### Hitung kofaktor

$$
C_{11} = (+1)(-2) = -2
$$

$$
C_{12} = (-1)(1) = -1
$$

$$
C_{13} = (+1)(0) = 0
$$


### Substitusi

$$
\det(A) = 0(-2) + 2(-1) + (-3)(0)
$$

$$
= -2
$$

**Hasil:**
$$
\det(A) = -2

## 4. Matriks 2×2

Diketahui:
$$
A =
\begin{bmatrix}
-7 & -5 \\
1 & 4
\end{bmatrix}
$$

---

### Determinan

$$
\det(A) = (-7)(4) - (-5)(1)
$$

$$
= -28 + 5 = -23
$$



### Minor

$$
M_{11} =
\begin{vmatrix}
4
\end{vmatrix} = 4
$$

$$
M_{12} =
\begin{vmatrix}
1
\end{vmatrix} = 1
$$

$$
M_{21} =
\begin{vmatrix}
-5
\end{vmatrix} = -5
$$

$$
M_{22} =
\begin{vmatrix}
-7
\end{vmatrix} = -7
$$



### Kofaktor

Gunakan:
$$
C_{ij} = (-1)^{i+j} M_{ij}
$$

$$
C_{11} = (+1)(4) = 4
$$

$$
C_{12} = (-1)(1) = -1
$$

$$
C_{21} = (-1)(-5) = 5
$$

$$
C_{22} = (+1)(-7) = -7
$$



### Matriks Kofaktor

$$
C =
\begin{bmatrix}
4 & -1 \\
5 & -7
\end{bmatrix}
$$



### Adjoin

$$ adj(A)= A^T $$
$$
Mengubah baris menjadi kolom dan kolom menjadi baris
\text{adj}(A) =
\begin{bmatrix}
4 & 5 \\
-1 & -7
\end{bmatrix}
$$



### Invers

$$
A^{-1} = \frac{1}{\det(A)} \cdot \text{adj}(A)
$$

$$
= \frac{1}{-23}
\begin{bmatrix}
4 & 5 \\
-1 & -7
\end{bmatrix}
$$


### Hasil Akhir

$$
A^{-1} =
\begin{bmatrix}
-\frac{4}{23} & -\frac{5}{23} \\
\frac{1}{23} & \frac{7}{23}
\end{bmatrix}
$$
